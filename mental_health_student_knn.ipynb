```python
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

# Encode seluruh dataset
encoder = OrdinalEncoder()
dataset_encoded = encoder.fit_transform(dataset[['gender','degree_level','academic_year','residential_status','campus_discrimination',
                                                 'sports_engagement', 'study_satisfaction','academic_pressure','financial_concerns',
                                                 'social_relationships','depression']])

# Gunakan nama kolum asal
dataset_encoded_df = pd.DataFrame(dataset_encoded, columns=['gender','degree_level','academic_year','residential_status','campus_discrimination',
                                                 'sports_engagement', 'study_satisfaction','academic_pressure','financial_concerns',
                                                 'social_relationships','depression'])

dataset_encoded_df['age'] = dataset['age']
dataset_encoded_df.head()
```

```python
dataset_encoded_df.info()
```

# Step 3: Split the data.¶¶

```python
dataset_encoded_df = dataset_encoded_df.drop(['degree_level', 'academic_year', 'residential_status', 'campus_discrimination',],axis=1)
dataset_encoded_df
```

```python
x = dataset_encoded_df.drop('depression',axis=1)
y = dataset_encoded_df['depression']
y = y.values.reshape(-1,1)
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.1, random_state=0)
x_train.shape, x_test.shape ,y_train.shape,y_test.shape
```

# Data scaling

```python
from sklearn.preprocessing import StandardScaler
XScalerFit = StandardScaler().fit(x_train)
YScalerFit = StandardScaler().fit(y_train)
# train data scaled
x_train_scaled = XScalerFit.transform(x_train)
# test data scaled
x_test_scaled = XScalerFit.transform(x_test)
```

```python
x_train_scaled
```

# Step 4: Build and train the model¶

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report

# Define KNN model
knn = KNeighborsClassifier()

# Define parameter grid
param_grid = {
    'n_neighbors': [1, 2, 3, 4, 5, 7, 8, 9, 10, 11, 12, 13, 14, 15],
    'metric': ['euclidean', 'manhattan', 'minkowski'],
    'weights': ['uniform', 'distance'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree', 'brute']
}


# Set up GridSearchCV
grid_search = GridSearchCV(estimator=knn, param_grid=param_grid, cv=5, scoring='accuracy', verbose=0)

# Fit the model
grid_search.fit(x_train_scaled, y_train)

best_model = grid_search.best_estimator_
# Best hyperparameters and evaluation
print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation Score:", grid_search.best_score_)
```

```python
from sklearn.metrics import accuracy_score
y_knn_train_best_pred = best_model.predict(x_train_scaled)# based on weights, make the prediction
y_knn_test_best_pred = best_model.predict(x_test_scaled)# based on weights, make the prediction

knn_best_train_acc = accuracy_score(y_train, y_knn_train_best_pred)# calculate the accuracy scor
knn_best_test_acc = accuracy_score(y_test, y_knn_test_best_pred)# calculate the accuracy scor
print('Train Accuracy:', round(knn_best_train_acc*100,2),'%')# print the accuracy score

print('Test Accuracy:', round(knn_best_test_acc*100,2),'%')# print the accuracy score
```

```python
from sklearn.neighbors import KNeighborsClassifier
training_accuracy = []
test_accuracy = []
parameter = []
k = range(1, 7)
metric = ['euclidean']
for n_k in k:
 clf = KNeighborsClassifier(n_neighbors=n_k, metric='euclidean',
weights='distance')
 clf.fit(x_train, y_train)
 training_accuracy.append(clf.score(x_train, y_train))
 test_accuracy.append(clf.score(x_test, y_test))
 parameter.append({'n_neighbors': n_k, 'metric': 'euclidean'})
```

```python
data_accuracy = pd.DataFrame.from_dict(parameter)# visualize the prediction result
data_accuracy['training_accuracy'] = training_accuracy
data_accuracy['test_accuracy'] = test_accuracy
data_accuracy
```

```python
fig = plt.figure(figsize=(5, 4))
ax1 = fig.add_subplot(1, 1, 1)
ax1.plot(data_accuracy['n_neighbors'],
data_accuracy['training_accuracy']*100, label='Training accuracy')
ax1.plot(data_accuracy['n_neighbors'], data_accuracy['test_accuracy']*100,
label='Test accuracy')
ax1.set_xlabel('K neighbours')
ax1.set_ylabel('Accuracy (%)')
plt.legend()
plt.show()
```

# Accuracy¶¶
